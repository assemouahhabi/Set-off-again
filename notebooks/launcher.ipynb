```python
%load_ext autoreload
%autoreload 2

# Detect if kaggle environment is being used
import os
import sys
import subprocess
from pathlib import Path

in_kaggle = Path("/kaggle/working").exists()

if in_kaggle:
    repo_dir = Path("/kaggle/working/Set-off-again")
    repo_url = "https://github.com/assemouahhabi/Set-off-again.git"
    
    # Clone the repository if it doesn't exist
    if not repo_dir.exists():
        subprocess.run(["git", "clone", repo_url, str(repo_dir)], check=True)
    else:
        # Pull the latest changes if the repository already exists
        subprocess.run(["git", "-C", str(repo_dir), "pull"], check=True)
    
    # Install the required packages from requirements.txt
    requirements_file = repo_dir / "requirements.txt"
    if requirements_file.exists():
        subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", str(requirements_file)], check=True)
    else:
        print(f"requirements.txt not found in {repo_dir}. Please ensure the file exists.")
    root_dir = repo_dir
else:
    #Local environment, set root_dir to the parent of the current working directory if it's "notebooks", otherwise use the current working directory
    root_dir = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
    
#add project root to python path to import modules from src
if str(root_dir) not in sys.path:
    sys.path.insert(0, str(root_dir))
    
print(f"Environment: {'Kaggle' if in_kaggle else 'Local'} | Root: {root_dir}")
```
